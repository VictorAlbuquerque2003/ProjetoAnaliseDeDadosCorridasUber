- Segunda pergunta de negócio: Qual é a taxa de conclusão das corridas?

```python
query = """
    SELECT 
        COUNT(*) FILTER(WHERE "Booking_Status"  = \'Completed\') / COUNT(*)::numeric 
    FROM ncr_ride_bookings
"""
```

```python
cursor.execute(query)
```

```python
porcentagem_corridas_concluidas = cursor.fetchall()
```

```python
float(porcentagem_corridas_concluidas[0][0] * 100)
```

    Cerca de 62% das corridas foram bem sucedidas.

- Terceira pergunta de negócio: Qual é a distância total percorrida e a distância média por corrida concluída?

```python
query = """
    SELECT 
        SUM("Ride_Distance") 
    FROM ncr_ride_bookings 
    WHERE "Booking_Status" = \'Completed\'
"""
```

```python
cursor.execute(query)
```

```python
distancia_total = cursor.fetchall()[0][0]
```

```python
distancia_total
```

```python
query = """
    SELECT 
        AVG("Ride_Distance") 
    FROM ncr_ride_bookings 
    WHERE "Booking_Status" = \'Completed\'
"""
```

```python
cursor.execute(query)
```

```python
distancia_media =  cursor.fetchall()[0][0]
```

```python
distancia_media
```

    A distância percorrida total das corridas concluídas foi de 2.398.883 KM e a distância média percorrida por corrida concluída foi de 26 KM.
